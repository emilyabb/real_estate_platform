```python
%sql

SELECT
  usage_metadata.serverless_compute_id,
  identity_metadata.run_as,
  SUM(usage_quantity) AS total_dbus
FROM system.billing.usage
WHERE billing_origin_product IN ('JOBS', 'DLT', 'INTERACTIVE')
  AND usage_metadata.serverless_compute_id IS NOT NULL
  AND usage_date >= date_add(now(), -30)
GROUP BY 1, 2
ORDER BY 3 DESC;
```

```python

# Trigger selective full refresh for a single table
from databricks.sdk import WorkspaceClient
import time

# Configuration
pipeline_id = "3ed566f9-0bd3-4d9f-9daa-1b91cadb0877"  # real_estate_new pipeline
fq_tables = [
    "silver_dev.opportunity_insights.social_capital_zip"
]
# Initialize client
w = WorkspaceClient()

# Start the selective full refresh
result = w.pipelines.start_update(
    pipeline_id=pipeline_id,
    full_refresh_selection=fq_tables
)
update_id = result.update_id
print("Update ID:", update_id)

# Wait for update to progress
time.sleep(60)

# Check full refresh update status
update = w.pipelines.get_update(
    pipeline_id=pipeline_id,
    update_id=update_id
)
print("State:", update.update.state)
print("Full refresh selection:", update.update.full_refresh_selection)
```
